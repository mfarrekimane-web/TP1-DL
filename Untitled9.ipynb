```python
import pandas as pd
from shapely import wkb

# 1) Catégories à garder, rangées par "famille"
familles = {
  "culture": ["history_museum","art_museum","science_museum","modern_art_museum",
              "civilization_museum","contemporary_art_museum","community_museum",
              "state_museum","childrens_museum","textile_museum","museum"],
  "patrimoine": ["monument","castle","palace","historic_site","historic_mission",
                 "historical_tour"],
  "nature_loisirs": ["national_park","state_park","botanical_garden","zoo",
                     "aquarium","amusement_park","water_park"],
  "plage": ["beach","beach_resort"],
  "souk": ["flea_market","farmers_market","public_market"],
  "stade": ["football_stadium","soccer_stadium","basketball_stadium",
            "track_stadium","rugby_stadium","tennis_stadium"],
}
cat_vers_famille = {c: f for f, liste in familles.items() for c in liste}

# 2) Filtrer
df["famille"] = df["cat_precise"].map(cat_vers_famille)
df["famille"] = df["famille"].fillna(
    df["basic_category"].map({"beach": "plage", "historic_site": "patrimoine"})
)
lieux = df[df["famille"].notna()].copy()

# 3) Nom, coordonnées
lieux["nom"] = lieux["names"].apply(lambda n: n["primary"] if n is not None else None)
def point(g):
    g = wkb.loads(g) if isinstance(g, (bytes, bytearray)) else g
    return g.x, g.y
lieux[["lon", "lat"]] = lieux["geometry"].apply(lambda g: pd.Series(point(g)))

# 4) Qualité minimale : un nom et une bonne confiance
lieux = lieux[lieux["nom"].notna() & (lieux["confidence"] >= 0.5)]

# 5) Résultat
print("Lieux gardés :", len(lieux))
print(lieux["famille"].value_counts())
print(lieux[["nom", "cat_precise", "famille", "lon", "lat"]].head(10))

# 6) Sauvegarde
lieux[["id","nom","cat_precise","famille","confidence","lon","lat"]].to_csv(
    "/content/drive/MyDrive/chouf_reco/lieux_visiteurs.csv", index=False)
```

```python
import numpy as np

lieux = lieux.reset_index(drop=True)

def haversine(lat1, lon1, lat2, lon2):
    # distance en km entre deux points
    R = 6371
    p1, p2 = np.radians(lat1), np.radians(lat2)
    dphi = p2 - p1
    dl = np.radians(lon2 - lon1)
    a = np.sin(dphi/2)**2 + np.cos(p1)*np.cos(p2)*np.sin(dl/2)**2
    return 2 * R * np.arcsin(np.sqrt(a))

def recommander(nom_ref, mode=1, rayon_km=30, n=10):
    # 1) lieu de référence
    trouve = lieux[lieux["nom"].str.contains(nom_ref, case=False, na=False)]
    if trouve.empty:
        return "Lieu introuvable : " + nom_ref
    ref = trouve.iloc[0]
    print(f"Référence : {ref['nom']} ({ref['famille']})")

    # 2) distance
    c = lieux[lieux["id"] != ref["id"]].copy()
    c["distance_km"] = haversine(ref["lat"], ref["lon"], c["lat"], c["lon"])
    total = len(c)
    c = c[c["distance_km"] <= rayon_km]
    print(f"Rejetés (trop loin) : {total - len(c)}")

    # 3) mode 1 = même famille / mode 2 = autre ambiance
    if mode == 1:
        c = c[c["famille"] == ref["famille"]]
    else:
        c = c[c["famille"] != ref["famille"]]

    if c.empty:
        return "Aucun résultat : essaie un rayon plus grand."

    # 4) score = proximité (70 %) + fiabilité des données (30 %)
    c["score"] = 0.7 * np.exp(-c["distance_km"] / 10) + 0.3 * c["confidence"]
    c = c.sort_values("score", ascending=False)

    # mode 2 : au maximum 3 lieux par famille, pour varier
    if mode == 2:
        c = c.groupby("famille").head(3).sort_values("score", ascending=False)

    return c.head(n)[["nom", "famille", "distance_km", "score"]].round(2)

# TEST
print(recommander("Volubilis", mode=1))
print()
print(recommander("Volubilis", mode=2))
```

```python
def recommander_id(id_ref, mode=1, rayon_km=30, n=10):
    ref = lieux[lieux["id"] == id_ref].iloc[0]
    print(f"Référence : {ref['nom']} ({ref['famille']})")

    c = lieux[lieux["id"] != id_ref].copy()
    c["distance_km"] = haversine(ref["lat"], ref["lon"], c["lat"], c["lon"])
    total = len(c)
    c = c[c["distance_km"] <= rayon_km]
    rejetes = total - len(c)

    if mode == 1:
        c = c[c["famille"] == ref["famille"]]
    else:
        c = c[c["famille"] != ref["famille"]]

    if c.empty:
        return f"Aucun résultat (trop loin : {rejetes}). Essaie un rayon plus grand."

    c["score"] = 0.7 * np.exp(-c["distance_km"] / 10) + 0.3 * c["confidence"]
    c = c.sort_values("score", ascending=False)
    if mode == 2:
        c = c.groupby("famille").head(3).sort_values("score", ascending=False)

    def raisons(r):
        liste = [f"à {r['distance_km']:.1f} km"]
        liste.append("même famille" if mode == 1 else "nouvelle ambiance")
        if r["confidence"] >= 0.8:
            liste.append("données fiables")
        return ", ".join(liste)

    c["raisons"] = c.apply(raisons, axis=1)
    print(f"Rejetés (trop loin) : {rejetes}")
    return c.head(n)[["nom", "famille", "distance_km", "score", "raisons"]].round(2)

VOLUBILIS = "bea77ed6-9601-4288-8fda-bc8bdff8e610"
print(recommander_id(VOLUBILIS, mode=1, rayon_km=60))
print()
print(recommander_id(VOLUBILIS, mode=2, rayon_km=60))
```

```python
# A) retirer les faux stades (clubs de tennis)
lieux = lieux[lieux["cat_precise"] != "tennis_stadium"].reset_index(drop=True)

# B) rayon et vitesse selon le déplacement
MODES = {"pied": (3, 4), "taxi": (15, 25), "excursion": (60, 50)}

def recommander_v2(id_ref, mode=1, deplacement="taxi", n=10):
    rayon, vitesse = MODES[deplacement]
    ref = lieux[lieux["id"] == id_ref].iloc[0]
    print(f"Référence : {ref['nom']} ({ref['famille']}) | déplacement : {deplacement}")

    c = lieux[lieux["id"] != id_ref].copy()
    c["distance_km"] = haversine(ref["lat"], ref["lon"], c["lat"], c["lon"])
    c = c[~((c["famille"] == ref["famille"]) & (c["distance_km"] < 0.3))]
    total = len(c)
    c = c[c["distance_km"] <= rayon]
    rejetes = total - len(c)

    c = c[c["famille"] == ref["famille"]] if mode == 1 else c[c["famille"] != ref["famille"]]

    if c.empty:
        return (f"Aucun résultat. Contrainte bloquante : distance max {rayon} km "
                f"({rejetes} lieux trop loin). Essaie 'taxi' ou 'excursion'.")

    # trajet estimé (détour de 30 %)
    c["trajet_min"] = (c["distance_km"] * 1.3 / vitesse * 60).round(0)

    c["score"] = 0.7 * np.exp(-c["distance_km"] / (rayon / 3)) + 0.3 * c["confidence"]
    c = c.sort_values("score", ascending=False)
    if mode == 2:
        c = c.groupby("famille").head(3).sort_values("score", ascending=False)

    return c.head(n)[["nom", "famille", "distance_km", "trajet_min", "score"]].round(2)

# TESTS
VOLUBILIS = "bea77ed6-9601-4288-8fda-bc8bdff8e610"
print(recommander_v2(VOLUBILIS, mode=1, deplacement="taxi"))
print()
print(recommander_v2(VOLUBILIS, mode=2, deplacement="excursion"))

# test en ville : Dar Jamai Museum (Meknès) à pied
DAR_JAMAI = lieux[lieux["nom"] == "Dar Jamai Museum"]["id"].iloc[0]
print()
print(recommander_v2(DAR_JAMAI, mode=2, deplacement="pied"))
```

```python
# A) retirer la plage mal placée
lieux = lieux[lieux["id"] != "be14a736-7ae6-4807-a020-03912b056720"].reset_index(drop=True)

# B) durée de visite estimée par famille (en heures)
DUREE_VISITE = {"culture": 1.5, "patrimoine": 1.0, "nature_loisirs": 2.0,
                "plage": 2.0, "souk": 1.0, "stade": 2.0}

# C) moteur avec durée disponible
def recommander_v3(id_ref, mode=1, deplacement="taxi", duree_h=3, n=10):
    rayon, vitesse = MODES[deplacement]
    ref = lieux[lieux["id"] == id_ref].iloc[0]
    print(f"Référence : {ref['nom']} | {deplacement} | {duree_h} h disponibles")

    c = lieux[lieux["id"] != id_ref].copy()
    c["distance_km"] = haversine(ref["lat"], ref["lon"], c["lat"], c["lon"])
    c = c[~((c["famille"] == ref["famille"]) & (c["distance_km"] < 0.3))]
    total = len(c)

    c = c[c["distance_km"] <= rayon]
    apres_distance = len(c)

    c = c[c["famille"] == ref["famille"]] if mode == 1 else c[c["famille"] != ref["famille"]]

    # temps total = aller-retour + visite
    c["trajet_min"] = (c["distance_km"] * 1.3 / vitesse * 60).round(0)
    c["visite_min"] = c["famille"].map(DUREE_VISITE) * 60
    c["total_min"] = 2 * c["trajet_min"] + c["visite_min"]
    avant_duree = len(c)
    c = c[c["total_min"] <= duree_h * 60]

    if c.empty:
        return {"resultats": 0,
                "trop_loin": total - apres_distance,
                "trop_long": avant_duree,
                "conseil": "Augmente la durée, le rayon, ou change de déplacement."}

    c["score"] = 0.7 * np.exp(-c["distance_km"] / (rayon / 3)) + 0.3 * c["confidence"]
    c = c.sort_values("score", ascending=False)
    if mode == 2:
        c = c.groupby("famille").head(3).sort_values("score", ascending=False)

    return c.head(n)[["nom", "famille", "distance_km", "trajet_min",
                      "visite_min", "total_min", "score"]].round(2)

# TESTS
DAR_JAMAI = lieux[lieux["nom"] == "Dar Jamai Museum"]["id"].iloc[0]
print(recommander_v3(DAR_JAMAI, mode=2, deplacement="pied", duree_h=3))
print()
print(recommander_v3(DAR_JAMAI, mode=2, deplacement="pied", duree_h=1))
```
